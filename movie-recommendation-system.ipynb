```python
movies = movies[['genres', 'id', 'keywords', 'title', 'overview', 'cast', 'year', 'vote_average','crew']]
```

```python
import ast

def convert(obj):
    L = []
    if isinstance(obj, str):
        obj = ast.literal_eval(obj)  
    for i in obj:
        L.append(i["name"])
    return L

movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['genres']
```

```python
movies.iloc[0]['cast']
```

```python
def actor(obj):
    L = []
    if isinstance(obj, str):
           obj = ast.literal_eval(obj)  
    for i in obj[:3]:  
        L.append(i["name"])
    return L

movies['cast'] = movies['cast'].apply(actor)
movies['cast']
```

```python
movies.iloc[0]['keywords']
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies['keywords']
```

```python
def fetch(obj):
    L = []
    for i in obj:
        if i["job"] == "Director":
            L.append(i["name"])
    return L

movies["crew"] = movies["crew"].apply(fetch)
```

```python
movies["overview"] = movies["overview"].apply(lambda x:x.split())
```

```python
movies['vote_average'] = pd.cut(movies['vote_average'], bins=[0, 3.3, 6.6, 10], labels=['lowrate','mediumrate','highrate'], include_lowest=True)
```

```python
movies.head()
```

```python
movies["genres"] = movies["genres"].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies["keywords"] = movies["keywords"].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies["cast"] = movies["cast"].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies["crew"] = movies["crew"].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
movies['vote_average'] = movies['vote_average'].astype(str)
```

```python
movies['tags'] = movies["genres"] + movies["keywords"] + movies["cast"] + movies["crew"] + movies["overview"] + movies["vote_average"].apply(lambda x:[x]) + movies["year"].apply(lambda x:[str(x)])
```

```python
movies.iloc[0]["tags"]
```

```python
df = movies[["id","title","tags"]]
```

```python
df.head(5)
```

```python
df["tags"] = df["tags"].apply(lambda x:" ".join(x))
```

```python
df["tags"] = df["tags"].apply(lambda x:x.lower())
```

```python
import nltk

from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
def shorten(text):
    L = []
    for i in text.split():
        L.append(ps.stem(i))

    return L

df['tags'] = df['tags'].apply(shorten)
```

```python
df["tags"] = df["tags"].apply(lambda x:" ".join(x))
```

```python
!pip install scikit-learn
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 5000, stop_words = 'english')
vectors = cv.fit_transform(df['tags']).toarray()
```

```python
vectors.shape
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similiarity = cosine_similarity(vectors)
```

```python
similiarity.shape
```

```python
def recommend(movie):
    ind = df[df['title'] == movie].index.values[0]
    print(ind)
    movie_list = sorted(list(enumerate(similiarity[ind])), reverse = True, key = lambda x: x[1])[1:7]
    print(movie_list)
    for i in movie_list:
        print(df.iloc[i[0]]['title'])
recommend('Avatar')   
```
